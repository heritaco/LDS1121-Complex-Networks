# Centralidad de Katz y PageRank

Cargamos las librerías

```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

Vamos a analizar las centralidades de varios ejemplos

## Ejemplo 1

```python
G=nx.DiGraph()
```

```python
stark_targaryen = ['Rickard','Lyarra','Aerys','Rhaella','Catelyn','Ned','Lyanna','Rhaegar','Elia','Viserys','Daenerys',
                   'Brandon', 'Benjen', 'Robb', 'Sansa', 'Arya', 'Bran', 'Rickon', 'Rhaenys', 'Aegon','Jon']

parents = [('Rickard','Ned'),('Rickard','Lyanna'),('Rickard','Brandon'),('Rickard','Benjen'),
           ('Lyarra','Ned'),('Lyarra','Lyanna'),('Lyarra','Brandon'),('Lyarra','Benjen'),
           ('Aerys','Rhaegar'),('Aerys','Viserys'),('Aerys','Daenerys'),
           ('Rhaella','Rhaegar'),('Rhaella','Viserys'),('Rhaella','Daenerys'),
           ('Catelyn','Robb'),('Catelyn','Sansa'), ('Catelyn','Arya'), ('Catelyn','Bran'), ('Catelyn','Rickon'),
           ('Ned','Robb'),('Ned','Sansa'), ('Ned','Arya'), ('Ned','Bran'), ('Ned','Rickon'),
           ('Rhaegar','Rhaenys'), ('Rhaegar','Aegon'),('Elia','Rhaenys'), ('Elia','Aegon'),
           ('Lyanna','Jon'),('Rhaegar','Jon')]

G.add_nodes_from(stark_targaryen)
G.add_edges_from(parents)

layout=nx.kamada_kawai_layout(G)
nx.draw(G,layout, with_labels=True)
```

### Centralidad de Katz

Usamos la función "katz_centrality" para calcular la centralidad de grado de los nodos.

```python
katz_centralidades=nx.katz_centrality(G)
katz_centralidades
```

Ordenando tenemos:

```python
dict(sorted(katz_centralidades.items(), key=lambda item:item[1], reverse=True))
```

Podemos notar que ahora 'Jon' es el nodo más importante.

### Centralidad de PageRank

Hacemos lo mismo con la centralidad de PageRank, usando la función "pagerank" obtenemos un diccionario, que ordenamos de mayor a menor.

```python
PR_centralidades=nx.pagerank(G)
PR_centralidades
```

```python
sorted(PR_centralidades.items(), key=lambda item:item[1], reverse=True)
```

Podemos observar que Jon sigue siendo el nodo más importante, sin embargo Robb bajó su centralidad y Rhaenys la aumentó.

#### Visualizando:

```python
katz_lista = [katz_centralidades[i] for i in G]

katz_sizes = np.array(katz_lista)

katz_colors = np.array([katz_centralidades[i] for i in G])
```

```python
plt.figure(figsize = [7,7] )

nx.draw(G,with_labels=True,node_size = 1000*katz_sizes,node_color = katz_colors)
```

```python
PR_lista = [PR_centralidades[i] for i in G]

PR_sizes = np.array(PR_lista)

PR_colors = np.array([PR_centralidades[i] for i in G])

nx.draw(G,with_labels=True,node_size = 2000*PR_sizes,node_color = PR_colors)
```

## Ejemplo 2

Utilicemos la red de colaboración científica.

```python
datos = pd.read_csv('ca-netscience.mtx', skiprows = 2, header = None,sep = ' ')
```

```python
aristas = datos[[0,1]].values
G1 = nx.Graph()
G1.add_edges_from(aristas)
```

Obtengamos las centralidades de Katz

```python
katz_centralidades_CA=nx.katz_centrality(G1,alpha=0.07) #Aquí fue necesario cambiar el parámetro de alpha
dict(sorted(katz_centralidades_CA.items(), key=lambda item:item[1], reverse=True))
```

#### Visualizando:

```python
katz_lista_CA = [katz_centralidades_CA[i] for i in G1]

katz_sizes_CA = np.array(katz_lista_CA)

katz_colors_CA = np.array([katz_centralidades_CA[i] for i in G1])

nx.draw(G1,node_size = 500*katz_sizes_CA,node_color = katz_colors_CA)
```

Hagamos lo mismo con la centralidad de PageRank

```python
PR_centralidades_CA=nx.pagerank(G1)
dict(sorted(PR_centralidades_CA.items(), key=lambda item:item[1], reverse=True))
```

#### Visualizando:

```python
PR_lista_CA = [PR_centralidades_CA[i] for i in G1]

PR_sizes_CA = np.array(PR_lista_CA)

PR_colors_CA = np.array([PR_centralidades_CA[i] for i in G1])

nx.draw(G1,node_size = 10000*PR_sizes_CA,node_color = PR_colors_CA)
```
